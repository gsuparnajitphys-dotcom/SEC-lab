### Problem 7

For the model used in introductory physics courses, a projectile thrown vertically at some initial velocity $v_i$ has position

$$
y(t) = y_i + v_i t - \frac{1}{2} g t^2
$$

where $g = 9.8\,\text{m/s}^2$.

Write a Python program that creates two lists: one containing time data (50 data points over 5 seconds) and the other containing the corresponding vertical position data for this projectile. The program should ask the user for the initial height $y_i$ and initial velocity $v_i$, and print $y(t)$ vs. $t$.


````python
#Date:03/09/2026
#Problem 7

import numpy as np

def projectile(yi,vi):
  yt=[]
  t=[]

  g=-9.8
  t0=np.linspace(0,5,50)

  for t1 in t0:
    t.append(t1)
    yt.append(yi+vi*t1+0.5*g*t1**2)

  return t , yt
````

# Write in python to calculate the value of $e^{2.31}$ using the following series

$$ e^x=\sum_0 ^{\infty} \frac{x^k}{k!}$$

#Correct upto 3 decimal places

````python
#Date:10/09/2026

import numpy as np
from scipy.special import factorial
import math



t=200
e=0

for i in range(0,t+1):
  e+=((2.31)**i)/(factorial(i, exact=False))

print(e)

````

````python
#Date:10/09/2026

import math
x=2.31
sum=0
term=1.0 #first term (k=0)
tolerence=0.0005
k=0

while abs(term)>=tolerence:
  sum=sum+term
  k=k+1
  term=term*x/(k) #next term

print(f"For x={x}; e^x ; sum ","=",format(sum,".3f"))
print(f"Number of terms used ={k}")
print(f"Actual value = {math.e**x}")
print(f"Error = {abs(math.e**x - sum)}")
````

# $\sin x = \sum_{k=0}^{\infty}(-1)^{k}\frac{x^{2k+1}}{(2k+1)!}$

````python
#Date:10/09/2026
import sympy as sp

x,k=sp.symbols("x,k")
def term(k):

  t0=((-1)**k)*((x**(2*k + 1))/sp.factorial(2*k+1))
  return t0
term(k)
````

````python
#Date:10/09/2026

term0=sp.lambdify((x,k),sp.simplify(term(k)),"numpy")

req0=sp.lambdify((x,k),sp.simplify(term(k+1)/term(k)),"numpy")

````

````python
#Date:10/09/2026

x=np.pi/4
k=0
sum0=0
term=term0(x,k)  #first term (k=0)
tolerence=0.00005



while abs(term)>=tolerence:
  sum0=sum0+term

  term=term*req0(x,k)  #next term
  k=k+1

print(f"For x={x}; sin(x) ; sum ","=",sum0)
print(f"Number of terms used ={k}")
print(f"Actual value = {np.sin(x)}")
print(f"Error = {abs(np.sin(x) - sum0)}")
````
